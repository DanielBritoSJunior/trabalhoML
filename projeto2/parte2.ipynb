## Eixo 1: Tratamento e Premissas  
* Pergunta: Se o dataset fosse altamente desbalanceado, por que a Acurácia seria uma métrica mentirosa para avaliar o sucesso do projeto?"


A acurácia pode enganar porque ela mostra apenas a quantidade total de acertos do modelo, sem levar em conta se ele está acertando justamente os casos mais importantes. No caso desse projeto de câncer de mama, imagine que o dataset tivesse 95% de tumores benignos e apenas 5% malignos. O modelo poderia classificar praticamente tudo como benigno e ainda assim teria 95% de acurácia. Mesmo com um número alto, ele falharia exatamente nos casos mais perigosos, que são os malignos.

Por isso, no projeto foram utilizadas outras métricas além da acurácia, como Recall e F1-Score. Elas ajudam a analisar melhor a capacidade do modelo identificar pacientes realmente doentes e reduzir os falsos negativos, que são os erros mais graves em um cenário médico.

* Pergunta: Para o KNN, como a falta de normalização dos dados afetaria o cálculo da distância entre os vizinhos?"


O KNN funciona comparando distâncias entre os dados para encontrar os vizinhos mais próximos. Quando os atributos possuem escalas muito diferentes, as variáveis com valores maiores acabam influenciando muito mais o cálculo da distância do que as outras.

Nesse dataset existem características como área, raio e textura do tumor, todas com escalas diferentes. Sem a normalização, uma variável com números muito altos poderia dominar o cálculo e fazer o modelo praticamente ignorar outras informações importantes.

Por isso foi utilizado o StandardScaler no código. O escalonamento deixa os atributos em uma escala parecida, permitindo que todas as características tenham influência mais equilibrada no treinamento do modelo


## Eixo 2: Arquitetura e Trade-offs  

* Pergunta: Nas Árvores de Decisão, como o parâmetro de profundidade máxima (max_depth) ajuda a controlar o Overfitting?"  


Nas Árvores de Decisão, o parâmetro max_depth ajuda a controlar o overfitting limitando o crescimento da árvore. Quando a árvore cresce demais, ela começa a criar regras muito específicas para os dados de treino e acaba “decorando” exemplos ao invés de aprender padrões mais gerais.

Ao definir uma profundidade máxima, o modelo fica mais simples e evita divisões desnecessárias causadas por ruídos ou exceções presentes no dataset. Isso melhora a capacidade de generalização do modelo para novos dados.


* Pergunta: "Por que o Random Forest costuma ser mais robusto a ruídos do que uma única Árvore de Decisão?"  


O Random Forest costuma ser mais robusto porque ele utiliza várias árvores de decisão ao mesmo tempo. Cada árvore é treinada com partes aleatórias dos dados e das variáveis, e no final a decisão acontece por votação.

Isso reduz bastante o impacto de ruídos e outliers, já que dificilmente todas as árvores vão cometer o mesmo erro ao mesmo tempo. Mesmo que algumas árvores aprendam padrões ruins, o conjunto tende a compensar isso.

Já uma única Árvore de Decisão é mais sensível aos dados e pode acabar sofrendo mais com overfitting, principalmente quando cresce demais. 

## Eixo 3: Interpretação de Performance  

* Pergunta: Analisando a Matriz de Confusão, o que é pior para o seu problema específico: um Falso Positivo ou um Falso Negativo? Justifique com base no negócio." 


Nesse problema de câncer de mama, o pior erro possível é o falso negativo. Isso acontece quando o modelo classifica um tumor maligno como benigno.

Em um cenário médico, esse erro pode atrasar o diagnóstico e o tratamento do paciente, aumentando o risco de agravamento da doença. Já o falso positivo, apesar de causar preocupação e exames adicionais, ainda mantém o paciente em acompanhamento médico.

Por isso, métricas como Recall são tão importantes nesse projeto, já que elas medem a capacidade do modelo identificar corretamente os casos realmente malignos. 

* Pergunta: Expliquem o que o F1-Score nos diz que a Acurácia não consegue mostrar."  


O F1-Score mostra o equilíbrio entre Precisão e Recall. Diferente da acurácia, ele não olha apenas para o total de acertos, mas também considera a qualidade desses acertos.

Isso é importante porque um modelo pode ter uma acurácia alta e ainda assim falhar justamente nos casos mais críticos. O F1-Score ajuda a identificar esse problema porque penaliza tanto falsos positivos quanto falsos negativos.

Por isso ele é muito utilizado em problemas médicos e em datasets desbalanceados, onde apenas a acurácia não é suficiente para avaliar o desempenho do modelo.
 

## Eixo 4: Decisão Estratégica  

* Pergunta: Considerando a  Navalha de Ockham, se a Regressão Logística teve 92% de acurácia e o SVM teve 93%, qual você recomendaria para produção em tempo real?"


Seguindo a ideia da Navalha de Ockham, a Regressão Logística seria uma escolha muito interessante nesse cenário. Apesar do SVM apresentar uma acurácia um pouco maior, a diferença é pequena e pode não justificar o aumento de complexidade.

A Regressão Logística é mais simples, mais rápida e mais fácil de interpretar. Em aplicações médicas, conseguir entender como o modelo chegou em determinada decisão é algo muito importante.

Além disso, modelos mais simples costumam ser mais leves computacionalmente e mais fáceis de manter em produção. Então, quando o desempenho é parecido, normalmente vale mais a pena utilizar a solução mais simples e estável.
